```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv('../data/GlobalWeatherRepository.csv')
```

```python
df
```

```python
alfa = df[['country', 'location_name', 'latitude', 'longitude', 'last_updated', 'temperature_celsius', 'condition_text', 'wind_kph', 'precip_mm', 'humidity', 'cloud', 'air_quality_gb-defra-index', 'sunrise', 'sunset']]
```

```python
alfa['last_updated'] = pd.to_datetime(alfa['last_updated'])
```

```python
alfa.info()
```

```python
alfa
```

```python
alfa['year'] = alfa['last_updated'].dt.year
alfa['month'] = alfa['last_updated'].dt.month
alfa['day'] = alfa['last_updated'].dt.day
alfa = alfa.drop(columns='last_updated')
```

```python
alfa['sr_dt'] = pd.to_datetime(alfa['sunrise'])
alfa['ss_dt'] = pd.to_datetime(alfa['sunset'])
```

```python
alfa['light_h'] = alfa['ss_dt'] - alfa['sr_dt']
```

```python
alfa
```

```python
# .dt.total_seconds() nos da los segundos, dividimos por 3600 para tener horas
alfa['sun_h'] = alfa['light_h'].dt.total_seconds() / 3600
alfa['sun_h'] = alfa['sun_h'].round(2)
```

```python
alfa.drop(columns=['sr_dt', 'ss_dt', 'light_h', 'sunrise', 'sunset'], inplace=True)
```

```python
alfa[['latitude', 'longitude']] = alfa[['latitude', 'longitude']].round(2)
```

```python
alfa.rename(columns={'location_name': 'capital', 'temperature_celsius': 'temp_C', 'air_quality_gb-defra-index': 'Air_poll'}, inplace=True)
```

```python
alfa['country_id'], uniques = pd.factorize(alfa['country'])
```

```python
alfa = alfa[['year', 'month', 'day', 'country_id', 'country', 'capital', 'latitude', 'longitude', 'wind_kph', 'precip_mm', 'humidity', 'cloud', 'Air_poll', 'sun_h', 'temp_C']]
alfa
```
